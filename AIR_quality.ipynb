```python
data=df.dropna()
data
```

```python
data.shape
```

```python
data['Solar'].dropna(inplace=True)
```

```python
data.drop_duplicates()
```

```python
from scipy import stats
import numpy as np
```

```python
z_scores = np.abs(stats.zscore(data['Ozone'].values))
print(z_scores)
```

```python
threshold = 3
outliers_z = np.where(z_scores > threshold)[0]

print(outliers_z)
print(len(outliers_z))
```

```python
# Identify the boolean mask for outliers
outlier_mask = z_scores > threshold

# Get the actual index labels of the outlier rows from the 'data' DataFrame
outlier_labels = data.index[outlier_mask]

# Drop these rows using their actual index labels and reset the index
df_remove = data.drop(index=outlier_labels).reset_index(drop=True)
display(df_remove)
```

```python
print(f'Number of outliers detected by Z-score (threshold=3) in bmi: {len(outliers_z)}')
df_z_removed = df.drop(index=outliers_z).reset_index(drop=True)
print('Shape before:', df.shape, 'Shape after Z-score removal:', df_z_removed.shape)
```

```python
Q1 = data['Ozone'].quantile(0.25)
Q3 = data['Ozone'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = data[(data['Ozone'] < lower_bound) | (data['Ozone'] > upper_bound)].index
print(f'Number of outliers detected by IQR in bmi: {len(outliers_iqr)}')
df_iqr_removed = data.drop(index=outliers_iqr).reset_index(drop=True)
print('Shape before:', data.shape, 'Shape after IQR removal:', df_iqr_removed.shape)
```
