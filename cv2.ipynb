```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# 1. Generate a blank image (Height, Width, Channels)
# A 3-channel image (BGR) of size 512x512
image = np.zeros((512, 512, 3), dtype=np.uint8)

# 2. Fill it with a color (OpenCV uses BGR format)
# This sets the entire image to Blue: (Blue=255, Green=0, Red=0)
image[:] = [255, 0, 0]

# 3. Save the generated image
cv2.imwrite('generated_image.jpg', image)

# 4. Display the image in a window
cv2_imshow(image)
# cv2.imshow('Generated Image', image) # Original line causing error
cv2.waitKey(0) # Wait for any key press to close
cv2.destroyAllWindows()
```
