```python
# A simple rule-based chatbot
def simple_chatbot(user_input):
    user_input = user_input.lower()

    if 'hello' in user_input or 'hi' in user_input:
        return 'Hello there! How can I help you today?'
    elif 'how are you' in user_input:
        return 'I am a chatbot, so I don\'t have feelings, but I\'m ready to assist you!'
    elif 'name' in user_input:
        return 'I am a simple rule-based chatbot.'
    elif 'bye' in user_input or 'goodbye' in user_input:
        return 'Goodbye! Have a great day!'
    elif 'help' in user_input:
        return 'I can answer basic questions. Try asking about my name or how I am.'
    else:
        return 'I\'m sorry, I don\'t understand that. Can you rephrase?'

# Let's test the chatbot
print(simple_chatbot('Hi'))
print(simple_chatbot('What is your name?'))
print(simple_chatbot('How are you doing?'))
print(simple_chatbot('Tell me something interesting'))
print(simple_chatbot('Goodbye'))
```
