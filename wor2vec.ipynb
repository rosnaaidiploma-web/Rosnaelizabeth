```python
# Import Libraries
from gensim.models import Word2Vec, FastText
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')

```

```python
# Sample Training Data
sentences = [
    "Machine learning is interesting",
    "Natural language processing is a part of AI",
    "Word embeddings help computers understand text",
    "Deep learning improves machine learning models",
    "FastText and Word2Vec are embedding techniques"
]

```

```python
# Preprocess Text
def preprocess_data(sentences):
    tokenized = []
    for sentence in sentences:
        words = word_tokenize(sentence.lower())
        tokenized.append(words)
    return tokenized

```

```python
# Word2Vec Function
def train_word2vec(data):

    model = Word2Vec(
        sentences=data,
        vector_size=100,
        window=5,
        min_count=1,
        workers=4
    )
    # Move the print statement before the return so it's reachable
    print("\nWord2Vec Model Trained Successfully")
    return model

# The following lines were causing the NameError as 'model' was not defined in the global scope
# and they should be handled in the main program logic, not at the global scope of this function definition.
# word = input("Enter word to find similar words: ")

# if word in model.wv:
#         print("\nSimilar Words:")
#         print(model.wv.most_similar(word))
# else:
#         print("Word not found in vocabulary")
```

```python
def input_sentence():

    text = input("\nEnter a sentence or paragraph:\n")

    data = preprocess(text)

    train_word2vec(data)
```

```python
# Preprocess the sample sentences
def preprocess_data(sentences):
    return [sentence.split() for sentence in sentences]
```

```python
def train_fasttext(data):

    model = FastText(
        sentences=data,
        vector_size=100,
        window=5,
        min_count=1,
        workers=4
    )

    print("\nFastText Model Trained Successfully")

    word = input("Enter word to find similar words: ")
    print(model.wv.most_similar(word))
```

```python
def main():

    data = preprocess_data(sentences)

    while True:

        print("\n--- Word Embedding Program ---")
        print("1. Train Word2Vec Model")
        print("2. Train FastText Model")
        print("3. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            train_word2vec(data)

        elif choice == "2":
            train_fasttext(data)

        elif choice == "3":
            print("Program Ended")
            break

        else:
            print("Invalid Choice")


```

```python
# Run Program
main()
```
