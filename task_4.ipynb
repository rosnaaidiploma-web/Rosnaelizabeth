```python
# ==============================
# Restaurant Ratings Analysis
# ==============================

# Step 1 - Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Step 2 - Load Dataset Function
def load_dataset():
    try:
        df = pd.read_csv("Dataset.csv")
        print("\nDataset Loaded Successfully!")
        print("Shape:", df.shape)
        print(df.head())
        return df
    except:
        print("Error loading dataset.")
        return None


# Step 3 - Show Dataset Info
def dataset_info(df):
    print("\nDataset Columns:")
    print(df.columns)

    print("\nDataset Info:")
    print(df.info())


# Step 4 - Rating Distribution Graph
def rating_distribution(df):

    if 'Aggregate rating' in df.columns:

        plt.figure(figsize=(8,5))
        sns.histplot(df['Aggregate rating'], bins=10, kde=True)

        plt.title("Distribution of Restaurant Ratings")
        plt.xlabel("Aggregate Rating")
        plt.ylabel("Number of Restaurants")

        plt.show()

    else:
        print("Aggregate rating column not found.")


# Step 5 - Most Common Rating Range
def common_rating_range(df):

    if 'Aggregate rating' in df.columns:

        rating_range = pd.cut(df['Aggregate rating'], bins=[0,1,2,3,4,5])

        most_common = rating_range.value_counts().idxmax()

        print("\nMost Common Rating Range:", most_common)

    else:
        print("Aggregate rating column not found.")


# Step 6 - Average Votes
def average_votes(df):

    if 'Votes' in df.columns:

        avg_votes = df['Votes'].mean()

        print("\nAverage Votes Received by Restaurants:", round(avg_votes,2))

    else:
        print("Votes column not found.")


# Step 7 - Votes Distribution Graph
def votes_graph(df):

    if 'Votes' in df.columns:

        plt.figure(figsize=(8,5))

        sns.histplot(df['Votes'], bins=20)

        plt.title("Votes Distribution")
        plt.xlabel("Votes")
        plt.ylabel("Restaurant Count")

        plt.show()

    else:
        print("Votes column not found.")


# Step 8 - Summary Report
def report(df):

    print("\n------ Analysis Report ------")

    if 'Aggregate rating' in df.columns:

        rating_range = pd.cut(df['Aggregate rating'], bins=[0,1,2,3,4,5])
        common = rating_range.value_counts().idxmax()

        print("Most Common Rating Range:", common)

    if 'Votes' in df.columns:

        avg_votes = df['Votes'].mean()

        print("Average Votes:", round(avg_votes,2))


# Step 9 - Main Menu (Switch Case)
def main():

    df = load_dataset()

    if df is None:
        return

    while True:

        print("\n========= MENU =========")
        print("1 - Dataset Info")
        print("2 - Rating Distribution")
        print("3 - Most Common Rating Range")
        print("4 - Average Votes")
        print("5 - Votes Distribution Graph")
        print("6 - Generate Report")
        print("7 - Exit")

        choice = input("Enter your choice: ")

        # Switch Case using if elif

        if choice == "1":
            dataset_info(df)

        elif choice == "2":
            rating_distribution(df)

        elif choice == "3":
            common_rating_range(df)

        elif choice == "4":
            average_votes(df)

        elif choice == "5":
            votes_graph(df)

        elif choice == "6":
            report(df)

        elif choice == "7":
            print("Program Ended")
            break

        else:
            print("Invalid Choice")


# Run Program
main()
```
