```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet
import sys

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python

# Initialize stemmer and lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

```

This code creates **(initializes)** two objects —


* **PorterStemmer(**) - for reducing words to their root form  

* **WordNetLemmatizer()** - for converting words to their meaningful base form , and their effect becomes visible when you pass input text to them and print the processed output.








```python
# Function for Stemming
def do_stemming(text):
    words = word_tokenize(text)
    stems = []
    for word in words:
        stems.append(stemmer.stem(word))
    return stems


# Function for Basic Lemmatization (default noun)
def do_lemmatization(text):
    words = word_tokenize(text)
    lemmas = []
    for word in words:
        lemmas.append(lemmatizer.lemmatize(word))
    return lemmas


# Function for Lemmatization with POS
def do_lemmatization_pos(text):
    words = word_tokenize(text)
    lemmas = []

```

 **T**hese functions take a text input, split it into words using *word_tokenize()*, then process each word using either **stemming** or **lemmatization**, store the results in a list, and return the processed words.


```python
def do_lemmatization_pos(text):
    words = word_tokenize(text)
    lemmas = []
    for word in words:
        print(f"\nChoose POS for '{word}':")
        print("1 - Noun")
        print("2 - Verb")
        print("3 - Adjective")
        print("4 - Adverb")

        pos_choice = input("Enter option: ")

        pos = wordnet.NOUN   # Default to Noun

        if pos_choice == "1":
            pos = wordnet.NOUN
        elif pos_choice == "2":
            pos = wordnet.VERB
        elif pos_choice == "3":
            pos = wordnet.ADJ
        elif pos_choice == "4":
            pos = wordnet.ADV
        # If pos_choice is not 1-4, pos remains wordnet.NOUN (the default)

        lemmas.append(lemmatizer.lemmatize(word, pos))

    return lemmas
```

This function **tokenizes** the input text into words, asks the user to choose the **Part of Speech (POS)** for each word, applies **lemmatization** based on the selected **POS (default noun)**, and **returns** the **list of lemmatized words**.

```python
# Main Menu Function
def main():

    print("Python Version:", sys.version)



    print("\n----- NLP Text Processing Program -----")
    paragraph = input("Enter a paragraph:\n")

    while True:
        print("\nSelect Option:")
        print("1 - Stemming")
        print("2 - Lemmatization (Default Noun)")
        print("3 - Lemmatization with POS")
        print("4 - Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            result = do_stemming(paragraph)
            print("\nStemming Result:")
            print(result)

        elif choice == "2":
            result = do_lemmatization(paragraph)
            print("\nLemmatization Result:")
            print(result)

        elif choice == "3":
            result = do_lemmatization_pos(paragraph)
            print("\nLemmatization with POS Result:")
            print(result)

        elif choice == "4":
            print("Program Ended.")
            break

        else:
            print("Invalid Option. Please try again.")


```



This main function displays the **Python version**, asks the **user to enter a paragraph**, then repeatedly shows a **menu to choose** stemming or lemmatization, processes the paragraph based on the **selected option**, prints the ***result***, and stops only when the **user selects Exit**.


```python

#Run the program
main()
```

When `main()` is executed:

 1. The program starts running by **displaying the Python version**

 2. Asking the user to  **enter a paragraph**, showing the menu options

 3. Then printing the processed result **(stemming or lemmatization**) based on the user's choice until **Exit** is selected.
