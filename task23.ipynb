```python
import pandas as pd

def load_data(file_path):
    """Loads the dataset and handles potential errors."""
    try:
        # Encoding latin-1 is used for this specific Zomato dataset
        df = pd.read_csv(file_path, encoding='latin-1')
        return df
    except Exception as e:
        print(f"Error: Could not load file. {e}")
        return None

def clean_data(df):
    """Cleans data and returns the processed DataFrame."""
    print("--- Starting Data Cleaning ---")
    df_clean = df.copy()
    # Filling missing Cuisines
    df_clean['Cuisines'] = df_clean['Cuisines'].fillna('Other')
    # Removing duplicates
    df_clean.drop_duplicates(inplace=True)
    return df_clean

def main():
    file_path = 'Dataset .csv'

    # List of steps to perform in order
    steps = ['load', 'clean', 'analyze', 'complete']

    data = None

    for step in steps:
        print(f"\nExecuting step: {step.upper()}")

        # 1. IF Statement to route logic
        if step == 'load':
            data = load_data(file_path)

            # 2. BREAK Condition: Exit if data loading failed
            if data is None:
                print("Critical Failure: Data is missing. Breaking loop.")
                break
            else:
                print(f"Success: Loaded {len(data)} rows.")

        elif step == 'clean':
            # Check if data exists before cleaning
            if data is not None:
                data = clean_data(data)
                print("Success: Data cleaned.")
            else:
                print("Error: No data found to clean.")
                break

        elif step == 'analyze':
            # Example Analysis: Check for high-rated restaurants
            if not data.empty:
                high_rated = data[data['Aggregate rating'] > 4.5]
                print(f"Analysis: Found {len(high_rated)} Excellent rated restaurants.")
            else:
                print("No data available for analysis.")
                break

        elif step == 'complete':
            print("--- All processes finished successfully ---")
            break # Final break to exit the loop

if __name__ == "__main__":
    main()
```
