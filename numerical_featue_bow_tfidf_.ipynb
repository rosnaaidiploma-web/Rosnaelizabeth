```python
import re
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

**re** is used for text cleaning using regular expressions

**CountVectorizer & TfidfVectorizer** from `sklearn.feature_extraction.text` are used to convert text into numerical features using Bag-of-Words and TF-IDF method

```python
def get_input():
    sentences = []
    print("Enter 5 Sentences:")
    for i in range(5):
        s = input(f"Sentence {i+1}: ")
        sentences.append(s)
    return sentences
```

This function **get_input()** collects 5 sentences from the user using a loop, stores them in a list, and returns the list.

```python
def clean_text(sentences):
    stopwords = ["is", "am", "are", "the", "a", "an", "and", "in", "on", "at"]
    cleaned = []

    for s in sentences:
        s = s.lower()
        s = re.sub(r'[^a-zA-Z ]', '', s)

        words = s.split()
        words = [word for word in words if word not in stopwords]

        cleaned.append(" ".join(words))

    return cleaned
```





* Removes numbers

* Removes punctuation

* Removes unwanted stopwords

* Joins words back into sentence





```python
def apply_bow(sentences):
    vectorizer = CountVectorizer()
    bow = vectorizer.fit_transform(sentences)
    return vectorizer.get_feature_names_out(), bow.toarray()


def apply_tfidf(sentences):
    vectorizer = TfidfVectorizer()
    tfidf = vectorizer.fit_transform(sentences)
    return vectorizer.get_feature_names_out(), tfidf.toarray()
```

**apply_bow()** returns the list of words and their *frequency matrix*

while **apply_tfidf()** returns the list of words and their TF-IDF *weight matrix*.

```python
sentences = get_input()
cleaned_sentences = clean_text(sentences)

print("\nCleaned Sentences:")
for s in cleaned_sentences:
    print(s)
```

*Now all cleaned sentences will be displayed in one statement as a list instead of printing line by line*

```python
choice = input("\nChoose Method (1-BoW, 2-TFIDF): ")

if choice == "1":
    words, result = apply_bow(cleaned_sentences)
    print("\nBoW Output:")
    print(words)
    print(result)

elif choice == "2":
    words, result = apply_tfidf(cleaned_sentences)
    print("\nTF-IDF Output:")
    print(words)
    print(result)

else:
    print("Invalid Choice")


print("\nUsing Switch Case")
switch_choice = int(input("Enter 1 for BoW, 2 for TF-IDF: "))

match switch_choice:
    case 1:
        words, result = apply_bow(cleaned_sentences)
        print("\nBoW Output:")
        print(words)
        print(result)

    case 2:
        words, result = apply_tfidf(cleaned_sentences)
        print("\nTF-IDF Output:")
        print(words)
        print(result)

    case _:
        print("Invalid Option")
```

Now the output is displayed in one print statement for each case using if statement and switch case statement.
