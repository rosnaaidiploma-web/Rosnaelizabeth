```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('Dataset .csv')

# 1. Basic Data Exploration
print("--- Dataset Overview ---")
print(df.info())
print("\n--- Summary Statistics ---")
print(df.describe())

# 2. Key Insights
# Find the top 5 most popular cuisines
top_cuisines = df['Cuisines'].value_counts().head(5)
print("\n--- Top 5 Cuisines ---")
print(top_cuisines)

# Calculate the average rating across all restaurants
avg_rating = df['Aggregate rating'].mean()
print(f"\nAverage Aggregate Rating: {avg_rating:.2f}")

# 3. Visualization: Distribution of Aggregate Ratings
plt.figure(figsize=(10, 6))
plt.hist(df['Aggregate rating'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Number of Restaurants')
plt.grid(axis='y', alpha=0.75)
plt.show()

# 4. Filter Example: High-rated restaurants with Table Booking
high_rated_booking = df[(df['Aggregate rating'] >= 4.5) & (df['Has Table booking'] == 'Yes')]
print("\n--- Sample of High-Rated Restaurants with Table Booking ---")
print(high_rated_booking[['Restaurant Name', 'City', 'Aggregate rating']].head())
```
